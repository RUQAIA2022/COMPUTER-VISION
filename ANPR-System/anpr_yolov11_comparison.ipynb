```python
from ultralytics import YOLO

# 1. Load the YOLOv11 nano pre-trained model for comparison
model_v11 = YOLO('yolo11n.pt')

# 2. Train the YOLOv11 model using the exact same dataset configuration
results_v11 = model_v11.train(
    data='data.yaml',    # Same dataset configuration file
    epochs=25,           # Same number of epochs for a fair comparison
    imgsz=640,           # Image size
    batch=16,            # Batch size
    name='yolov11_anpr_comparison'  # Specific folder for YOLOv11 results
)

# 3. Evaluate YOLOv11 performance on the unseen test set
metrics_v11 = model_v11.val(data='data.yaml', split='test')

# 4. Run predictions on test images to visualize YOLOv11 results
results_pred = model_v11.predict(
    source='data/images/test',
    conf=0.25,
    save=True
)

print("YOLOv11 training, evaluation, and predictions completed successfully!")
```
