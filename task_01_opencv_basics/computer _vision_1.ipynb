```python
import cv2
import matplotlib.pyplot as plt

# 1. Read the image first
image = cv2.imread("data/cat_image.jpeg")

# 2. Convert colors and display the comparison
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(image)
axes[0].set_title("BGR (opencv)")
axes[0].axis("off")

axes[1].imshow(image_rgb)
axes[1].set_title("RGB (Matplotlib)")
axes[1].axis("off")

plt.show()
```

```python
print(image.shape)
```

```python
height, width, channels = image.shape
print(f"Height {height} pixels")
print(f"Width {width} pixels")
print(f"channels {channels}")
```

```python
(480,680,3)
```

```python
total_pixels = height * width
print(f"Total Pixels {total_pixels}")
print(f"Total Pixels in all channels {total_pixels * channels}")
```

```python
blue_channel = image_rgb[:, :, 0]
green_channel = image_rgb[:, :, 1]
red_channel = image_rgb[:, :, 2]

fig, axes = plt.subplots(1, 4, figsize=(16, 4))

axes[0].imshow(image_rgb)
axes[0].set_title("Original image")
axes[0].axis("off")

axes[1].imshow(blue_channel)
axes[1].set_title("Blue Channel")
axes[1].axis("off")

axes[2].imshow(green_channel)
axes[2].set_title("Green Channel")
axes[2].axis("off")

axes[3].imshow(red_channel)
axes[3].set_title("Red Channel")
axes[3].axis("off")

plt.show()
```

```python
print("Full image shape", image.shape)
print("Blue channel shape", blue_channel.shape)
print("Green channel shape", green_channel.shape)
print("Red channel shape", red_channel.shape)
```

```python
resized_image = cv2.resize(image_rgb, (240, 240))

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(image_rgb)
axes[0].set_title("Original image 360 x 360")
axes[0].axis("off")

axes[1].imshow(resized_image)
axes[1].set_title("Resized image 240 x 240")
axes[1].axis("off")

plt.show()
```

```python
print("Original", image_rgb.shape)
print("Resized Image", resized_image.shape)
```

```python
crop_image = image_rgb[50:360, 50:260]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(image_rgb)
axes[0].set_title("Original image 360 x 360")
axes[0].axis("off")

axes[1].imshow(crop_image)
axes[1].set_title("Cropped image")
axes[1].axis("off")

plt.show()
```
